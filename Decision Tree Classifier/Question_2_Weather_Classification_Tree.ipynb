# Question 2 – Classification Tree: Weather and Going Out

Predict whether a person will **Go Out** or **Stay In** based on the weather.

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt

data = pd.DataFrame({
    'Weather': ['Sunny', 'Sunny', 'Sunny', 'Rainy', 'Rainy', 'Rainy', 'Sunny', 'Rainy'],
    'Go_Out': ['Go Out', 'Go Out', 'Go Out', 'Stay In', 'Stay In', 'Stay In', 'Go Out', 'Stay In']
})

data
```

### Convert the weather categories into numbers

```python
data['Weather_Code'] = data['Weather'].map({'Sunny': 0, 'Rainy': 1})

X = data[['Weather_Code']]
y = data['Go_Out']

model = DecisionTreeClassifier(max_depth=1, criterion='gini', random_state=42)
model.fit(X, y)

print("Tree trained successfully.")
```

```python
plt.figure(figsize=(10, 6))
plot_tree(
    model,
    feature_names=['Weather'],
    class_names=['Go Out', 'Stay In'],
    filled=True
)
plt.show()
```

### Prediction

The classification rule learned from the data is:

- **Sunny → Go Out**
- **Rainy → Stay In**

```python
new_data = pd.DataFrame({'Weather_Code': [0, 1]})

predictions = model.predict(new_data)

for weather, prediction in zip(['Sunny', 'Rainy'], predictions):
    print(weather, "->", prediction)
```
